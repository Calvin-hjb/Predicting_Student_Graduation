```python
# for IEM data
for i in range(1, 9):
    # Load the Excel file
    file_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Trial data\\ClassificationS{i}.xlsx'
    df = pd.read_excel(file_path)
    
    # Splitting Data and Exporting to Excel
    X = df.iloc[:, :-1]
    y = df.iloc[:, -1]
    y.name = df.columns[-1]  # Preserve label column name
    
# 102: 0-113
# 103: 113-227
# 104: 227-337
# 105: 337-440
# 106: 440-542
# 107: 542-644
# 108: 644-734
    
    start_index = 0
    end_index = 113
    
    # Split Test and Train Set
    X_test = X.iloc[start_index:end_index,:] 
    y_test = y.iloc[start_index:end_index]
        
    # Create X_test and X_train using array slicing
    X_train = pd.concat((X.iloc[:start_index,:],X.iloc[end_index:,:]), axis = 0)
    y_train = pd.concat((y.iloc[:start_index],y.iloc[end_index:]), axis = 0)
    

    # Combined the Sliced data
    # Concatenate y to x 
    TrainData = pd.concat((X_train, y_train), axis=1)
    TestData = pd.concat((X_test, y_test), axis = 1)
    
    # Export train and test sets to Excel
    train_excel_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Trial data\\Train\\No102_S{i}.xlsx'
    test_excel_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Trial data\\Test\\Test102_S{i}.xlsx'
    TrainData.to_excel(train_excel_path, index=False)
    TestData.to_excel(test_excel_path, index=False)
    print(f"Processed and exported data for S{i}")
```

```python
# for Chemistry data
for i in range(1, 9):
    # Load the Excel file
    file_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Chemistry\\Data_S{i}.xlsx'
    df = pd.read_excel(file_path)
    
    #Data cleaning
    df = df.drop(columns = ['課名','班別','PR','先修_1','先修_2','暑修_1','暑修_2','修課次數'])
        # encoding the feature column
    replace_col = {'M' : 1,
                    'F' : 2,
                    '大學甄選入學個人申請' : 3,
                    '大學甄選入學繁星推薦' : 4,
                    '大學考試入學' : 5,
                    '延畢' : 1,
                    '退畢' : 1,
                    '畢業' : 0}
    df[1] = df[1].replace(replace_col)
    df[6] = df[6].replace(replace_col)
    df[7] = df[7].replace(replace_col)
    
    # Extracting data
    X = df.iloc[1:, 1:-1].values # [1:, 0:-1] if including Student and course name
    y = df.iloc[1:869, -1].values

    y = np.where(y == 1, 1, 0)
    # Reshape to (num_samples, height, width, channels)
    X = X.reshape(-1, 111, 8, 1) # (-1, 111, 9, 1) if including Student and course name  
    X = pd.DataFrame(X.reshape(868, -1))
        
        # Sort the data according to entry years 102,103,...,108
    X = X.sort_values(by=1) # by=2 if including Student and course name
    y = X[6] # take the collumn X[6] as the "real" target
    
    # Remove unnecessary column
        # column 2 = grade of the student at that time
        # column 3 = year of graduation
        # column 4 = semester of graduation
        # column 7 = graduation score
    X = X.drop(columns = [1,2,3,4,6,7]) # [3,4,5,8] if including Student and course name
    X

    # Removing NaN and all Zero column
        # Get column names where all values are NaN
    nan_columns_list = X.columns[X.isna().all()].tolist()
        # remove column with all Nan values
    X = X.dropna(axis =1)
        # Find columns where all values are 0
    non_zero_columns = (X != 0).any(axis=0)

        # Keep only columns with at least one non-zero value
    X = X.loc[:, non_zero_columns]

    # Make sure the index of the X and y are still the same after cleaning the data
    Data = np.hstack((X, y.reshape(-1, 1)))
    X = Data[0:, 0:-1]
    y = Data[0:869, -1]
    
    # Splitting Data and Exporting to Excel
    
# 102: 0-116 (116)
# 103: 116-235 (119)
# 104: 235-351 (116)
# 105: 351-468 (117)
# 106: 468-579 (111)
# 107: 579-694 (115)
# 108: 694-868 (174)
    
    start_index = 694
    end_index = 868
    
    # Convert to Pandas DataFrame/Series if needed

    X = pd.DataFrame(X)
    y = pd.Series(y)

    # Split Test and Train Set
    X_test = X.iloc[start_index:end_index,:] 
    y_test = y.iloc[start_index:end_index]
        
    # Create X_test and X_train using array slicing
    X_train = pd.concat((X.iloc[:start_index,:],X.iloc[end_index:,:]), axis = 0)
    y_train = pd.concat((y.iloc[:start_index],y.iloc[end_index:]), axis = 0)
    

    # Combined the Sliced data
    # Concatenate y to x 
    TrainData = pd.concat((X_train, y_train), axis=1)
    TestData = pd.concat((X_test, y_test), axis = 1)
    
    # Export train and test sets to Excel
    train_excel_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Chemistry\\Train\\No108_S{i}.xlsx'
    test_excel_path = f'D:\\AI Warning\\3_CNN\\Data and Classification Model\\Classification\\Chemistry\\Test\\Test108_S{i}.xlsx'
    TrainData.to_excel(train_excel_path, index=False)
    TestData.to_excel(test_excel_path, index=False)
    print(f"Processed and exported data for S{i}")
```
