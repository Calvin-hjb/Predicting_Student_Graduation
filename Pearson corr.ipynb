# Heat Map

```python
cor1 = X_1.corr()

cor2 = X_2.corr()

cor3 = X_3.corr()

cor4 = X_4.corr()

cor5 = X_5.corr()

cor6 = X_6.corr()

cor7 = X_7.corr()

cor8 = X_8.corr()
```

```python
cor1
```

```python
import numpy as np

# List of correlation matrices for each semester
cor_matrices = [cor1, cor2, cor3, cor4, cor5, cor6, cor7, cor8]

for sem_idx, cor in enumerate(cor_matrices, start=1):
    perfect_corr = np.where((cor > 0.99999999999) & (np.eye(cor.shape[0]) == 0))

    # Use a set to store unique feature pairs
    feature_pairs = set()
    for i, j in zip(*perfect_corr):
        pair = tuple(sorted([cor.index[i], cor.columns[j]]))  # Sort to remove duplicates
        feature_pairs.add(pair)

    # Display results
    print(f"Features with 1.0 correlation in Semester {sem_idx}:")
    for pair in feature_pairs:
        print(pair)
    print()  # Blank line for readability
```

```python
# List of correlation matrices from semester 1 to 8
cor_matrices = [cor1, cor2, cor3, cor4, cor5, cor6, cor7, cor8]

for i, cor in enumerate(cor_matrices, start=1):  
    # Find feature pairs with correlation of exactly 1.0 (excluding self-correlation)
    perfect_corr = np.where((cor < -0.9) & (np.eye(cor.shape[0]) == 0))

    # Extract feature names
    feature_pairs = [(cor.index[x], cor.columns[y]) for x, y in zip(*perfect_corr)]

    # Display results for the current semester
    print(f"\nFeatures with -1.0 correlation in Semester {i}:")
    if feature_pairs:
        for pair in feature_pairs:
            print(pair)
    else:
        print("No perfect correlations found.")
```
