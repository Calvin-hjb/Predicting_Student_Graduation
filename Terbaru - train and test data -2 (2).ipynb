```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches


# Combine the predictions and confidence probabilities across semesters
all_predictions = np.concatenate([y_pred1, y_pred2, y_pred3, y_pred4, y_pred5, y_pred6, y_pred7, y_pred8])
all_probabilities = np.concatenate([yscore_1, yscore_2, yscore_3, yscore_4, yscore_5, yscore_6, yscore_7, yscore_8])

# Create an array to indicate the semester for each data point
semester_labels = (
    ['Semester 1'] * len(y_pred1) + 
    ['Semester 2'] * len(y_pred2) + 
    ['Semester 3'] * len(y_pred3) +
    ['Semester 4'] * len(y_pred4) +
    ['Semester 5'] * len(y_pred5) +
    ['Semester 6'] * len(y_pred6) +
    ['Semester 7'] * len(y_pred7) +
    ['Semester 8'] * len(y_pred8) 
)

# Convert semester labels to numerical values for plotting (1, 2, 3)
semester_indices = (
    [1] * len(y_pred1) + 
    [2] * len(y_pred2) + 
    [3] * len(y_pred3) +
    [4] * len(y_pred4) +
    [5] * len(y_pred5) +
    [6] * len(y_pred6) +
    [7] * len(y_pred7) +
    [8] * len(y_pred8) 
)

# Plot the prediction confidence across semesters
plt.figure(figsize=(12, 8))
colors = ['blue' if pred == 0 else 'red' for pred in all_predictions]

# Use a scatter plot to show confidence over semesters
plt.scatter(semester_indices, all_probabilities, c=colors, edgecolor='k', s=100, alpha=0.7)
# Create custom legend handles
blue_patch = mpatches.Patch(color='blue', label='Prediction 0')
red_patch = mpatches.Patch(color='red', label='Prediction 1')
plt.legend(handles=[blue_patch, red_patch], loc='upper right')

plt.title('Prediction Confidence Across Semesters 1 to 8')
plt.xlabel('Semester')
plt.ylabel('Confidence Probability')
plt.xticks([1, 2, 3, 4, 5, 6,7,8], ['Semester 1', 'Semester 2', 'Semester 3', 'Semester 4', 'Semester 5', 'Semester 6', 'Semester 7', 'Semester 8'])  # Label x-axis for each semester
plt.axhline(y=0.5, color='gray', linestyle='--', linewidth=0.8)  # Reference line at 0.5
plt.grid(True)
plt.show()
```
