```python
df1 = pd.read_excel('D:\\AI Warning\\3_CNN\\Data and Classification Model\\Matching 108\\102-108\\classification_combined.xlsx', sheet_name="S8")
```

```python
# Extract features into numpy array X

df1 = df1.drop(columns = ['課名','班別','PR','先修_1','先修_2','暑修_1','暑修_2'])
X = df1.iloc[1:, 1:-1].values

# Extract target variable into numpy array y
y = df1.iloc[1:735, -1] 
y = y.dropna()

print(X.shape)
print(y.shape)
```

# Remove zero column

```python
# (X != 0).any(axis=0) --> this locate columns which has all zero value within them
# X.loc[:,(X != 0).any(axis=0)] --> this remove the columns that have all zero values 
NonZero = (X != 0).any(axis=0)
X = X.loc[:, NonZero]
```
